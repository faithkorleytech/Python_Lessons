```python
# Day 2: Heart Rate Analysis
heart_rates = [72, 75, 71, 120, 73, 74]

# 1. Calculate the average
average_bpm = sum(heart_rates) / len(heart_rates)

# 2. Display the status report
print(f"Analysis of {len(heart_rates)} heart rate samples:")
print(f"Average: {average_bpm:.1f} BPM")

# 3. Logic Challenge
if average_bpm > 80:
    print("⚠️ Status: High Heart Rate Alert!")
else:
    print("✅ Status: Normal")
```
